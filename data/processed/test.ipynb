```python
# Pull just the U.S. Senate race and pivot: counties as rows, candidates as columns
raw_96 = pd.read_excel("../raw/ark_gen_election_data/ark_gen_96.xls", header=None)

# Row 0 holds the county names (cols 2+); col 0 holds race titles / candidate names
counties = raw_96.iloc[0, 2:]

# Find the "U.S. SENATE" header row; candidates run until the next blank row
start = raw_96.index[raw_96[0].astype(str).str.strip() == "U.S. SENATE"][0] + 1
end = raw_96.index[(raw_96.index > start) & raw_96[0].isna()][0]

senate = raw_96.iloc[start:end, 2:]
senate.index = raw_96.iloc[start:end, 0].str.strip()   # candidate names
senate.columns = counties

senate_96 = senate.T.drop(index="TOTAL").astype(int)
senate_96.index.name = "County"
senate_96.columns.name = "Candidate"
senate_96
```
